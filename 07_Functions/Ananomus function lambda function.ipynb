```python
nums= [3,2,4,8,4,6,2,9]
evens= list(filter(lambda n: n%2==0,nums))
odds= list(filter(lambda n: n%2==1,nums))
print(evens)
print(odds)
```

```python
nums= [3,2,4,8,4,6,2,9]
evens= list(filter(lambda n: n+2,nums))
odds= list(map(lambda n: n+2,nums))
print(evens)
print(odds)
```
